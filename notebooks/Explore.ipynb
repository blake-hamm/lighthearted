```python
class InitProfile(dspy.Signature):
    """Build out inital profile for user."""
    reason: str = dspy.InputField(desc="reason for interacting with agent")
    goals: list[str] = dspy.InputField(desc="some current goals")
    suggested_goals: list[str] = dspy.OutputField(desc="list of suggested goals")
    questions: list[str] = dspy.OutputField(desc="list of follow up questions")
```

```python
reason = "I want to improve my mental health"
goals = ["Become an independent freelancer", "Build a local community", "Spend more time outdoors"]

profile = dspy.ChainOfThought(InitProfile)
profile(reason=reason, goals=goals)
```
